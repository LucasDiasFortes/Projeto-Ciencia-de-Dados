Este projeto tem como objetivo analisar os acidentes de trânsito com vítimas fatais registrados nas rodovias federais brasileiras, utilizando os dados abertos disponibilizados pela Polícia Rodoviária Federal (PRF). A análise busca identificar padrões relacionados aos dias da semana, horários, causas dos acidentes e diferenças entre os estados brasileiros.

**Objetivo geral**

Analisar os acidentes com vítimas fatais registrados nas rodovias federais brasileiras entre 2020 e 2025, buscando identificar padrões temporais e as principais causas associadas às mortes.


**Perguntas que o projeto pretende responder**

**Pergunta 1:** Em que dia da semana e horário os acidentes com mortes mais acontecem?

**Pergunta 2:** Quais causas mais levam a mortes, e isso muda por estado?

**Critérios de sucesso**

O projeto será considerado satisfatório se conseguir identificar os dias e horários com maior ocorrência de acidentes fatais, apresentar as principais causas relacionadas às mortes e demonstrar diferenças entre os estados brasileiros por meio de análises e visualizações compreensíveis.
