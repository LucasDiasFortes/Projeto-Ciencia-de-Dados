**Análise inicial dos acidentes com mortes**

Na base de dados referente ao ano de 2020, foram identificados 4.525 acidentes que registraram pelo menos uma pessoa morta. Ao somar a coluna mortos, foi identificado um total de 5.292 pessoas mortas.

Essa distinção é importante para o projeto, pois permite analisar tanto a quantidade de acidentes fatais quanto o número total de vítimas fatais.

```python
df['classificacao_acidente'].value_counts()
```

**Distribuição dos acidentes por classificação**

A análise da coluna classificacao_acidente identificou três categorias: acidentes com vítimas feridas, acidentes sem vítimas e acidentes com vítimas fatais.

Foram registrados 47.456 acidentes com vítimas feridas, 11.602 sem vítimas e 4.525 com vítimas fatais. Também foram identificados dois registros sem classificação preenchida.

Essa análise permite conhecer a distribuição geral dos acidentes e confirmar a quantidade de ocorrências classificadas como fatais na base de 2020.

```python
df[df['mortos'] > 0]['dia_semana'].value_counts()
```

**Acidentes com mortes por dia da semana**

A análise dos acidentes que registraram pelo menos uma morte demonstrou que o domingo apresentou a maior quantidade de ocorrências, totalizando 977 acidentes. Em seguida, aparecem sábado, com 806, e sexta-feira, com 665.

A quarta-feira apresentou a menor quantidade, com 493 acidentes.

Os resultados indicam uma concentração maior de acidentes com mortes nos finais de semana, especialmente aos domingos. Essa observação será aprofundada com a análise dos demais anos da pesquisa.

```python
df[df['mortos'] > 0]['horario'].value_counts().head(10)
```

**Acidentes com mortes por horário**

A análise dos horários dos acidentes com mortes demonstrou que 19h foi o horário com maior quantidade de ocorrências, totalizando 107 acidentes. Os horários de 18h30 e 20h aparecem em seguida, ambos com 93 ocorrências.

Também foi observada uma concentração de acidentes entre 18h e 20h30, indicando maior frequência de ocorrências nesse intervalo durante o ano de 2020.

```python
df[df['mortos'] > 0]['causa_acidente'].value_counts().head(10)
```

**Principais causas dos acidentes com mortes**

A análise das causas dos acidentes que registraram pelo menos uma morte identificou a falta de atenção à condução como a causa mais frequente, com 1.111 ocorrências. Em seguida, aparecem a desobediência às normas de trânsito pelo condutor, com 765, e a velocidade incompatível, com 579.

Também se destacaram a falta de atenção do pedestre, com 524 ocorrências, e a ingestão de álcool, com 297.

Os resultados indicam que fatores relacionados ao comportamento dos condutores e pedestres estão entre as causas mais frequentes dos acidentes com mortes registrados em 2020.

```python
df_fatais = df[df['mortos'] > 0]

df_fatais.groupby('uf')['mortos'].sum().sort_values(ascending=False)
```

**Distribuição de mortes por estado**

A análise da quantidade de pessoas mortas por estado demonstrou que Minas Gerais apresentou o maior número, com 717 mortes, seguido pelo Paraná, com 526, e pela Bahia, com 507.

Santa Catarina registrou 382 mortes e Pernambuco, 318. Por outro lado, os menores números foram observados no Amapá, com 5 mortes, Acre, com 13, e Amazonas, com 16.

Os resultados demonstram diferenças na distribuição de mortes entre os estados brasileiros. Entretanto, será necessário aprofundar a análise para verificar se as principais causas dos acidentes também variam de acordo com o estado.

```python
df_fatais.groupby(['uf', 'causa_acidente'])['mortos'].sum().reset_index()
```

```python
top_causas_estado = (
    df_fatais.groupby(['uf', 'causa_acidente'])['mortos']
    .sum()
    .reset_index()
    .sort_values(['uf', 'mortos'], ascending=[True, False])
    .groupby('uf')
    .head(3)
)

top_causas_estado
```

```python
top_causas_estado.head(15)
```

```python
principais_causas = (
    top_causas_estado
    .groupby('uf')
    .head(1)
)

principais_causas['causa_acidente'].value_counts()
```

**Comparação das principais causas por estado**

A comparação das principais causas associadas às mortes em cada unidade federativa demonstrou que a falta de atenção à condução ocupou a primeira posição em 19 dos 27 estados analisados.

A desobediência às normas de trânsito pelo condutor foi a principal causa em quatro estados, enquanto a velocidade incompatível ocupou essa posição em dois. A ingestão de álcool e a falta de atenção do pedestre foram as principais causas em um estado cada.

Os resultados indicam que a falta de atenção à condução apresenta predominância nacional na base analisada em 2020, mas também existem diferenças entre os estados.
